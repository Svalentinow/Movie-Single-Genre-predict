```python
new_df['Genre'] = new_df['Genre'].astype(str).str.strip()

# 2. Create a mask where the string is strictly a single alphabetical word
single_word_mask = new_df['Genre'].str.isalpha()

# 3. Apply the mask to get your clean rows
single_genre_df = new_df[single_word_mask]
```

```python
single_genre_df.Genre.value_counts().head(40)
```

```python
print(single_genre_df.Plot.isna().sum())
print(single_genre_df.Plot.duplicated().sum())
```

```python
duplicated_plots_df = single_genre_df[single_genre_df.Plot.duplicated(keep=False)]

# Sort by the Plot column so duplicates appear next to each other
duplicated_plots_df.sort_values(by='Plot')
```

```python
single_genre_df = single_genre_df.drop_duplicates(subset=['Plot'])
print(single_genre_df.Plot.duplicated().sum())
```

```python
single_genre_df.Genre.value_counts().head(10)
```

```python
top_genres = single_genre_df['Genre'].value_counts().head(10).index.tolist()
print(top_genres)
```

```python
filtered_df = single_genre_df[single_genre_df['Genre'].isin(top_genres)]

# 2. Group by genre and cap each category at a maximum of 500 rows
capped_df = filtered_df.groupby('Genre', group_keys=False).apply(
    lambda x: x.sample(n=min(len(x), 112), random_state=42)
) 

# 3. Check your newly balanced distribution
print(capped_df['Genre'].value_counts())
print(f"\nTotal rows left to scrape: {len(capped_df)}")
```

```python
capped_df.to_csv("filtered_movie_plots_4000.csv", index=False)
```

```python
df = pd.read_csv("wiki_movie_plots_deduped.csv")
df = df[~(df['Genre'] == 'unknown')]
df['Genre'].value_counts().head(10)
```

```python
df['Genre'] = df['Genre'].astype(str).str.strip()


# 2. Create a mask where the string is strictly a single alphabetical word
single_word_mask = df['Genre'].str.isalpha()

# 3. Apply the mask to get your clean rows
single_genre_df = df[single_word_mask]

single_genre_df = single_genre_df.drop_duplicates(subset=['Plot'])
print(single_genre_df.Plot.duplicated().sum())
```

```python
top_genres = single_genre_df['Genre'].value_counts().head(10).index.tolist()
print(top_genres)
```

```python
single_genre_df.Genre.value_counts().head(10)
```

```python
filtered_df2 = single_genre_df[single_genre_df['Genre'].isin(top_genres)]

# 2. Group by genre and cap each category at a maximum of 500 rows
capped2_df = filtered_df2.groupby('Genre', group_keys=False).apply(
    lambda x: x.sample(n=min(len(x), 400), random_state=42)
)

# 3. Check your newly balanced distribution
print(capped2_df['Genre'].value_counts())
```
